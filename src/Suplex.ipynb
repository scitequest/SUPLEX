```python
from suplex_elements.GUI import GUI
import ipywidgets as widgets
import logging
from suplex_elements.LogDisplay import LogDisplay
from modules import ModuleLoader as modules
```

```python
# Set up logging
logView = widgets.HTML()
logging.basicConfig(
    level = logging.INFO,
    format = "%(asctime)s - %(levelname)s - %(message)s",
    handlers=[logging.StreamHandler()]
    #handlers = [LogDisplay(logView)]
)
logger = logging.getLogger(__name__)

# Load Modules
modules = modules.loadOperationModules()

# Container for all elements of the GUI
content = widgets.VBox()

# Needed to allow dynamic updates to widgets
output = widgets.Output()

display(content, output)
gui = GUI(modules, content, logView)
```
