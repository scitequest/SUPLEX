```python
from ui.gui import GUI
import ipywidgets as widgets
import logging
import os

from ui.elements.popup_log_system import PopupLogger, PopupLogHandler

from suplex_modules.module_loader import load_modules
from suplex_modules.module_constants import SEARCH_MODE, DEFAULT_SEARCH_DEPTH
from ui.styles.scheme_loader import get_schemes
from ui.styles.schemes.default_scheme import DEFAULT_SCHEME
```

```python
# Set up logging
popup_logger = PopupLogger(max_messages=500, min_log_level = logging.INFO)
popup_handler = PopupLogHandler(popup_logger)

logging.basicConfig(
    level = logging.DEBUG,
    format = "%(message)s",
    handlers = [popup_handler]
)
logger = logging.getLogger(__name__)

modules = load_modules(max_depth = DEFAULT_SEARCH_DEPTH, mode = SEARCH_MODE.RECURSIVE)

schemes = get_schemes()
if not schemes:
    schemes[DEFAULT_SCHEME.NAME] = {DEFAULT_SCHEME}

#width = "800px"
#height = "600px"

#if "VOILA_APP_IP" in os.environ or "VOILA_BASE_URL" in os.environ:
#    width = "100vw"
#    height = "100vh"

# Container for all elements of the GUI
content = widgets.Box(layout=widgets.Layout(
                        display="grid", # Tolerates placement of widgets on top of one another
                        grid_template_rows="1fr",
                        grid_template_columns="1fr",
                    ))
content.add_class("gui-content-container")

# Needed to allow dynamic updates to widgets
output = widgets.Output()

display(content, output)
gui = GUI(modules, schemes, content, popup_logger)
gui.build_gui()
```
