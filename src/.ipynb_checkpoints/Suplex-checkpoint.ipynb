````python
# Jupyter Notebook Code zum Erstellen der Test-Struktur

## Zelle 1: Import und Setup

```python
import os
from pathlib import Path

# Basis-Verzeichnis für Tests
base_dir = Path("C:/Users/KaiKr/OneDrive/Python/Projects/suplex/tests")

# Verzeichnisstruktur definieren
structure = {
    "": [  # Root-Verzeichnis (tests/)
        "__init__.py",
        "conftest.py",
        "pytest.ini",
        "requirements_test.txt",
        "README_tests.md",
        "test_mocks.py"
    ],
    "unit": [
        "__init__.py",
        "test_martin_adapter.py",
        "test_metadata.py",
        "test_project.py",
        "test_roi_elements.py",
        "test_sample.py",
        "test_search_directories.py",
        "test_constants.py",
        "test_dataset.py",
        "test_general_parameters.py",
        "test_holder.py"
    ],
    "integration": [
        "__init__.py",
        "test_adapter_integration.py",
        "test_validation_integration.py"
    ],
    "notebooks": [
        "__init__.py",
        "test_notebooks.py"
    ]
}
```

## Zelle 2: Verzeichnisse und Dateien erstellen

```python
# Erstelle die Verzeichnisstruktur
def create_test_structure(base_path, structure_dict):
    """Erstellt die komplette Test-Verzeichnisstruktur mit leeren Dateien."""
    
    # Erstelle Basis-Verzeichnis
    base_path.mkdir(exist_ok=True)
    print(f"✓ Erstellt: {base_path}/")
    
    # Durchlaufe die Struktur
    for subdir, files in structure_dict.items():
        # Erstelle Unterverzeichnis wenn nötig
        if subdir:
            dir_path = base_path / subdir
            dir_path.mkdir(exist_ok=True)
            print(f"✓ Erstellt: {dir_path}/")
        else:
            dir_path = base_path
        
        # Erstelle alle Dateien im Verzeichnis
        for filename in files:
            file_path = dir_path / filename
            file_path.touch()  # Erstellt leere Datei
            print(f"  ✓ Datei: {file_path}")
    
    print("\n✅ Test-Struktur erfolgreich erstellt!")

# Führe die Erstellung aus
create_test_structure(base_dir, structure)
```

## Zelle 3: Struktur visualisieren

```python
def show_tree(directory, prefix="", is_last=True):
    """Zeigt die Verzeichnisstruktur als Baum an."""
    if prefix == "":
        print(directory.name + "/")
        prefix = ""
    
    contents = list(directory.iterdir())
    contents.sort(key=lambda x: (x.is_file(), x.name))
    
    for i, path in enumerate(contents):
        is_last_item = i == len(contents) - 1
        current_prefix = "└── " if is_last_item else "├── "
        print(prefix + current_prefix + path.name + ("/" if path.is_dir() else ""))
        
        if path.is_dir():
            extension = "    " if is_last_item else "│   "
            show_tree(path, prefix + extension, is_last_item)

# Zeige die erstellte Struktur
print("\n📁 Erstellte Test-Struktur:\n")
show_tree(base_dir)
```

## Zelle 4: Validierung

```python
# Validiere, dass alle Dateien erstellt wurden
def validate_structure(base_path, structure_dict):
    """Überprüft, ob alle Dateien und Verzeichnisse erstellt wurden."""
    all_good = True
    missing = []
    
    for subdir, files in structure_dict.items():
        dir_path = base_path / subdir if subdir else base_path
        
        # Prüfe Verzeichnis
        if not dir_path.exists():
            missing.append(f"Verzeichnis: {dir_path}")
            all_good = False
            continue
        
        # Prüfe Dateien
        for filename in files:
            file_path = dir_path / filename
            if not file_path.exists():
                missing.append(f"Datei: {file_path}")
                all_good = False
    
    if all_good:
        print("✅ Alle Dateien und Verzeichnisse wurden erfolgreich erstellt!")
    else:
        print("❌ Folgende Elemente fehlen:")
        for item in missing:
            print(f"   - {item}")
    
    return all_good

# Führe Validierung aus
validate_structure(base_dir, structure)
```

## Zelle 5: Statistiken anzeigen

```python
# Zeige Statistiken
def show_statistics(base_path):
    """Zeigt Statistiken über die erstellte Struktur."""
    total_files = 0
    total_dirs = 0
    
    for root, dirs, files in os.walk(base_path):
        total_files += len(files)
        total_dirs += len(dirs)
    
    print(f"\n📊 Statistiken:")
    print(f"   - Verzeichnisse: {total_dirs + 1}")  # +1 für base_dir
    print(f"   - Dateien: {total_files}")
    print(f"   - Gesamt: {total_dirs + 1 + total_files} Elemente")

show_statistics(base_dir)
```

## Zelle 6: Optional - Struktur löschen

```python
# Optional: Funktion zum Löschen der Struktur
def remove_test_structure(base_path):
    """Löscht die erstellte Test-Struktur."""
    import shutil
    
    if base_path.exists():
        response = input(f"⚠️  Wirklich '{base_path}' löschen? (ja/nein): ")
        if response.lower() == 'ja':
            shutil.rmtree(base_path)
            print(f"✅ '{base_path}' wurde gelöscht.")
        else:
            print("❌ Abgebrochen.")
    else:
        print(f"ℹ️  '{base_path}' existiert nicht.")

# Auskommentieren um zu nutzen:
# remove_test_structure(base_dir)
```

## Komplettes Notebook als eine Zelle:

```python
import os
from pathlib import Path
import shutil

# Definiere die komplette Struktur
def create_complete_test_structure():
    """Erstellt die komplette Test-Struktur mit einem Befehl."""
    
    base_dir = Path("tests")
    
    # Struktur definieren
    structure = {
        "": ["__init__.py", "conftest.py", "pytest.ini", "requirements_test.txt", 
             "README_tests.md", "test_mocks.py"],
        "unit": ["__init__.py", "test_martin_adapter.py", "test_metadata.py", 
                 "test_project.py", "test_roi_elements.py", "test_sample.py", 
                 "test_search_directories.py", "test_constants.py", "test_dataset.py", 
                 "test_general_parameters.py", "test_holder.py"],
        "integration": ["__init__.py", "test_adapter_integration.py", 
                        "test_validation_integration.py"],
        "notebooks": ["__init__.py", "test_notebooks.py"]
    }
    
    # Erstelle alles
    base_dir.mkdir(exist_ok=True)
    
    for subdir, files in structure.items():
        dir_path = base_dir / subdir if subdir else base_dir
        if subdir:
            dir_path.mkdir(exist_ok=True)
        
        for filename in files:
            (dir_path / filename).touch()
    
    print("✅ Test-Struktur wurde erstellt!")
    
    # Zeige Baum
    print("\n📁 Struktur:")
    for root, dirs, files in os.walk(base_dir):
        level = root.replace(str(base_dir), '').count(os.sep)
        indent = ' ' * 2 * level
        print(f"{indent}{os.path.basename(root)}/")
        subindent = ' ' * 2 * (level + 1)
        for file in sorted(files):
            print(f"{subindent}{file}")

# Ausführen
create_complete_test_structure()
```
````
